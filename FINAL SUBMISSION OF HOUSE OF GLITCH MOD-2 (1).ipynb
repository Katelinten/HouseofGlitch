# ARTIST STATEMENT 




Thinking about auto-theory, where one uses the self to draw out ideas that generalize certain concepts or systems [of thinking]; I couldn't help but use my own experiences for this digital project. For instance, what does it mean to be Black, a Woman, an American, a Born-digital youth? The byproduct of the intersectionality within these identities is being a <<double>> outsider, this means i’ve been glitching throughout my entire life. 
 
The idea of repetition compulsion is when you put yourself in situations that caused you past trauma with the hopes of fixing it this *time* around. This project is helping me confront the notion that I've been going full circle my whole life – but what if I am just stuck in a loop? <<helix>>.
 
To summarize the affect of this experience: Being me isn't as aggravating, as much as it is confusing as h*ll. And in being me, I mean what does it mean to be always already knowing? To clarify, the experience of living in the first and third-person simultaneously, being haunted and residing in a dream-space, the double outsider. The rememberer. 
 
I wanted this poem to be disorienting but wholly aware and hopeful. The feeling of being almost there–wherever there is– but constantly interrupted and imposed. The feeling of *finally* escaping but later finding you are not entirely free of the memories and impulse this experience has inculcated <<Helix>>. I would theorize that being aware of my identities and how they affect my life choices and life chances is the only way I can experience joy, dream, and conjure alternative identities and realities for myself through play and craft. The only way I can experience freedom while being unfree. 
This project is inspired by multiple Black writers and creatives, as well as sourcing critical race theory to ground the work. Throughout the poem, I also grab inspiration from artists in the Black community like Solange, who dives deep in the questions asked in the poem.  I also use urban and southern Black American vernacular throughout this project. 
 
My work starts with Du Bois' *The Souls of Black Folk*  and the theory of 'Double Consciousness'. DuBois claims Black Americans are gifted with second sight because we see everything through a veil: we see what we see and how other people see us seeing. So we have to see oursleves and also how people perceive us constantly. What does living in that doubled land, gifted with a second sight mean? It is important to note that this gift is a condition of oppression, so what does it mean again to make sense and use that middle – nowhere, everywhere – space? This gift is also a curse and means we see things differently, which pertains to spatial transversal to dream-space dimensions where one can speculate on their future itself? 
 
 
I am also thinking about Baldwin's non-fiction, *Collected essays* with the hero's journey. Thinking of the common story structure of 'the heroes journey' where when attempting to go home, the protagonist finds the journey fundamentally changes them so much they can no longer 'return' or finally return home for the *first* time as themselves. Maybe the 'changed' us is our real home? [figure 1 & 5 ]
 In the essay, 'The discovery of what it means to be an American' and to the hero's journey story structure, Baldwin's journey throughout Paris changes him so much he can finally return *home* for the first home as himself, a Negro American. Baldwin wants to be an American more than anything. Yet, he realizes he can't be one without first being anti-Black: internalizing notions of inferiority that prevent him from enjoying Bessie Smith and Watermelon. And secondly, extreme nihilism: The love of Blackness and your people leading you to be pessimistic. Baldwin states there is a rootlessness to Black American identity and experience.
Concerning experiences of anti-Blackness I am also thinking about Toni Morrisons *Beloved*, and how the timeless experiences of chattel slavery through re-memories have changed the circular nature of time from phenomena – the moment – to deep time –the Anthropocene or plantationocene – but only as symptoms and reproducers of social relations from slavery. 
    
The hero's journey is not circular, but a spiral; a helix; Baldwin demonstrates a full circle, yes, but it is not concluded or a feedback loop, it is a 360. Baldwin is exactly where he started but we new acquired knowledge and confidence in his path [figure 2-3]. Time and Space: Space-Time are two sides of the same coin. A moment in time is space, and a place is a moment in time [Figure 4]. Time keeps moving progressively; When we feel like we are in the same place, it is a different time when we feel like we are repeating events because we are in the same space.
Thinking about the rootlessness of Black American experiences due to the social dysfunction of plantation slavery, being imposed being imposed as Toni Morrison states in *Sula*, to "a second hand lonely" from the racially haunted space, the United States, and it's "half-buried" past. I identify with the circular concept of time and space, and the mental and emotional sequences of memories and consciousness as a spiral; going up or down [2-4]. Figure 1 and figure 5 describe a loop, figures 2-4 illustrate the third dimension of a loop; a helix/spiral.

Going back to the theoretical approach to the list categories, I want to hone in on light source. 

For lightsource I started out listing literal  forces of light like ‘sunlight’, then I moved towards light that glows in the dark like moonlight, flashlights, street lights – as a Black American child, my life was bounded by the street lights, it told me when it was time to go home.

 But then I wanted to expand this term ‘light source” scientifically in relation to things that glow in the dark; How melanin actually absorbs light sources like UV rays to protect against cellular damage. 

Then, I thought about light sources metaphorically, pertaining to light skin and the products that produce lighter complexions from make up to skin ‘care’ products: Black American slangs for people who have light skin are ‘yellow bone’ they are also referred to as ‘light bright’ and ‘high yella (yellow)’.

I am also thinking about concealer and how it’s supposed to brighten up the center of your face as well as foundation too many shades lighter than your actual skin tone which people do as a symptom of internalized anti-Blackness.

 Then finally, turmeric and lemon juice [recipes] which are used to brighten up natural skin tones. This is often to cater to the globalization of colorism in Black and and non-Black POC communities which translates proximity to whiteness with value and and therefore ‘beauty’, which i also added as a light source.I thought about light sources literally being lighter skin tone but also how the appearance of a lighter complexions makes your life more *light-weight*.


    I then thought of light souces theoretically: ‘anti-blackness’. Whiteness is a structure for domination, and it is constantly imagined and reimagined always in comparison and opposition to Black people in the Americas.
    In "Whiteness as Property," Cheryl Harris documents how owning Whiteness in the West allows the ability to own property and all the economic, social, and political fortunes that come from that membership. Non-Black people can be owners of Whiteness or aspire to Whiteness through anti-black and indigenous violence. Thinking of Whiteness historically, it came to replace property in the Americas. To clarify, globally, when many countries allowed the ownership of property to define citizenship, the Americas pioneered using Whiteness as a marker for citizenship and belonging to land. This not only shows how Whiteness has historically been treated as property, therefore an asset, but also touches back on why even oppressed people who may be eligible to access Whiteness, side with whiteness time and time again rather than communities that may hold the best interest for their socioeconomic class. 
    
So, knowing that Anti-Blackness is a energy souce  and causality for the power of whiteness and white supremacy, I added that to the list as well.
